```python
import torch
import numpy as np
import torch.nn as nn
from ModelUtils import *
from PackingUtils import *
from ModelCrossAttention import *
from train_model import get_dataloader
from tqdm import tqdm
# sklearn CrossEntropy loss
# from sklearn.
```

```python
model_path = "checkpoints/model_10.pth"
```

```python
train_loader, test_loader = get_dataloader("dataset_map", batch_size=8)

device = torch.device('cuda')

model = BPP_Model(
	image_size=(100, 100),
	patch_size=2,
	embed_dim=8,
	num_heads=4,
	hidden_dim=16,
	num_layers=3,
	encoded_dim=512,
	batch_size=8,
	encoder_mlp_dims=[256, 256],
	decoder_mlp_dims=[1024, 2048, 4096],
	output_size=(100, 100, 2),
)
```

```python
model.load_state_dict(torch.load(model_path, map_location=device))
model = model.to(device)
model.eval()
```

```python
pred_list = []
label_list = []
with torch.no_grad():
    progress_bar = tqdm(enumerate(test_loader))
    for i, (hm, isz, feas, label) in progress_bar:
        hm, isz, feas = hm.to(device), isz.to(device), feas.to(device)
        label = label.to(device)
        pred = model(hm, isz, feas)
        pred_list.append(pred.detach().cpu().numpy())
        label_list.append(label.detach().cpu().numpy())
        if i > 5:
            break
```
